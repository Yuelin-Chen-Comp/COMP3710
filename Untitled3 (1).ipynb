```python
# environment setting
import torch
import numpy as np
print ("PyTorch Version:", torch.__version__)
device = torch.device('cuda'
                      if torch.cuda.is_available() else 'cpu')
```

```python
# grid for computing image, subdivide the space
X, Y = np.mgrid[-4.0:4:0.01, -4:4:0.01] # generate the range of x axis and y axis from -4 to 4, interval is 0.01

x = torch.Tensor(X) # load into PyTorch tensors
y = torch.Tensor(Y)

x = x.to(device) # transfer to the GPU device
y = y.to(device)
```

```python
# compute Gassian Function
# where r = x^2 + y^2
z = torch.exp(-(x ** 2 + y ** 2) / 2.0)
```

```python
# plot
import matplotlib.pyplot as plt

plt.imshow(z.cpu().numpy()) # Updated
plt.tight_layout()
plt.show()
```

```python
# Task1.3.1: change the Gaussian function into a 2D sine
sine_wave = torch.sin(x * 3.0 + y * 3.0)

plt.imshow(sine_wave.cpu().numpy())
plt.tight_layout()
plt.show()
```

```python
# Task1.3.2: multiply both Gaussian and the sine, get gabor
gabor = z * sine_wave

plt.imshow(gabor.cpu().numpy())
plt.tight_layout()
plt.show()
```

```python
Y, X = np.mgrid[-1.5:1.5:0.005, -1.5:1.5:0.005]
x = torch.Tensor(X)
y = torch.Tensor(Y)
z = torch.complex(x, y)
zs = z.clone()
ns = torch.zeros_like(z)

# transfer to the GPU device
z = z.to(device)
zs = zs.to(device)
ns = ns.to(device)

# mandelbrot set
for i in range (200):
  # compute the new valies of z: z^2 + z
  zs_ = zs * zs + z

  # have we deverged with this new value?
  not_diverged = torch.abs(zs_) < 4.0

  # update variables to compute
  ns += not_diverged
  zs[not_diverged] = zs_[not_diverged]

# plot
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(16, 10))

def processFractal(a):
  a_cyclic = (6.28 * a / 20.0).reshape(list(a.shape) + [1])
  img = np.concatenate([10 + 20 * np.cos(a_cyclic),
                        30 + 50 * np.sin(a_cyclic),
                        155 - 80 * np.cos(a_cyclic)], 2)
  img[a == a.max()] = 0
  a = img
  a = np.uint8(np.clip(a, 0, 255))
  return a

plt.imshow(processFractal(ns.cpu().numpy()))
plt.tight_layout()
plt.show()

```

 difference between Mandelbrot and Julia
 z0 of Mandelbrot is 0, z0 of Julia will change with axis change(z0 = x +iy)
 c of Mandelbrot will change with axis change(c = x +iy), c of Julia is a fix number(c = -0.7 + 0.27015i)

```python
Y, X = np.mgrid[-1.5:1.5:0.005, -1.5:1.5:0.005]

x = torch.Tensor(X)
y = torch.Tensor(Y)

# ------------------------------------------------------------------
# the core chage of Julia
# 1. z_0 depends on axis x, y
# 2. c is a fixed negative constant
# ------------------------------------------------------------------
zs = torch.complex(x, y).to(device)  # z_0 = x + i*y

# choose a classic c: -0.7 + 0.27015i
c_real = torch.tensor(-0.7)
c_imag = torch.tensor(0.27015)
c = torch.complex(c_real, c_imag).to(device)

ns = torch.zeros_like(zs, dtype=torch.float32).to(device)

for i in range(200):
  zs = zs * zs + c  # notice, use c
  not_diverged = torch.abs(zs) < 4.0
  ns += not_diverged.type(torch.float32)

plt.imshow(ns.cpu().numpy())
plt.tight_layout()
plt.show()


```

Part 3 Newton Fractal

```python
"""
COMP3710 - Lab 1: Fractal Generation using PyTorch
Fractal Choice: Newton Fractal for the complex polynomial f(z) = z^3 - 1

Mathematical Principle:
Newton-Raphson iteration in the complex plane:
    z_{n+1} = z_n - f(z_n) / f'(z_n)
            = z_n - (z_n^3 - 1) / (3 * z_n^2)
            = (2 * z_n^3 + 1) / (3 * z_n^2)

Roots of z^3 - 1 = 0:
    r1 = 1
    r2 = -0.5 + i * sqrt(3)/2
    r3 = -0.5 - i * sqrt(3)/2

Hardware Acceleration:
    All complex numbers are stored in PyTorch tensors and computed
    in parallel across all grid pixels on GPU / CPU without manual loops.
"""

import matplotlib.pyplot as plt
from matplotlib.colors import hsv_to_rgb
import numpy as np
import torch

# -------------------------------------------------------------------------
# Step 1: Computational Device Configuration
# -------------------------------------------------------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Executing computations on device: {device}")

# -------------------------------------------------------------------------
# Step 2: Grid Generation on the Complex Plane [-2.0, 2.0] x [-2.0, 2.0]
# -------------------------------------------------------------------------
# Define a high-resolution grid using NumPy
step_size = 0.003
Y, X = np.mgrid[-2.0:2.0:step_size, -2.0:2.0:step_size]

# Convert NumPy coordinate grids to PyTorch tensors and send to target device
x = torch.tensor(X, dtype=torch.float32, device=device)
y = torch.tensor(Y, dtype=torch.float32, device=device)

# Construct complex tensor representing initial positions z_0
z = torch.complex(x, y)

# -------------------------------------------------------------------------
# Step 3: Define Theoretical Roots of f(z) = z^3 - 1
# -------------------------------------------------------------------------
roots = [
    torch.complex(
        torch.tensor(1.0, device=device), torch.tensor(0.0, device=device)
    ),
    torch.complex(
        torch.tensor(-0.5, device=device),
        torch.tensor(np.sqrt(3) / 2.0, dtype=torch.float32, device=device),
    ),
    torch.complex(
        torch.tensor(-0.5, device=device),
        torch.tensor(-np.sqrt(3) / 2.0, dtype=torch.float32, device=device),
    ),
]

# -------------------------------------------------------------------------
# Step 4: Iteration Setup and State Trackers
# -------------------------------------------------------------------------
max_iterations = 40
tolerance = 1e-4  # Convergence threshold

# Track which root each point converges to (1, 2, or 3)
root_indices = torch.zeros(z.shape, dtype=torch.int64, device=device)

# Track the number of iterations required to reach convergence
converged_iterations = torch.zeros(
    z.shape, dtype=torch.float32, device=device
)

# Active mask identifying points that have not yet converged
mask_active = torch.ones(z.shape, dtype=torch.bool, device=device)

# -------------------------------------------------------------------------
# Step 5: Newton-Raphson Iteration Loop (Tensor Parallelism)
# -------------------------------------------------------------------------
for i in range(1, max_iterations + 1):
  # Compute derivative-based update formula: z_{n+1} = (2*z^3 + 1) / (3*z^2)
  z_sq = z * z
  z = (2.0 * z * z_sq + 1.0) / (3.0 * z_sq)

  # Check convergence against all three theoretical roots
  for idx, root in enumerate(roots):
    distance = torch.abs(z - root)
    is_converged = (distance < tolerance) & mask_active

    # Record root index and iteration count
    root_indices[is_converged] = idx + 1
    converged_iterations[is_converged] = i

    # Update active mask to ignore already converged points
    mask_active = mask_active & (~is_converged)

# -------------------------------------------------------------------------
# Step 6: Advanced Visualisation using HSV Color Space Mapping
# -------------------------------------------------------------------------
# Transfer computed tensors back to CPU for rendering
root_idx_np = root_indices.cpu().numpy()
iters_np = converged_iterations.cpu().numpy()

# Normalize iteration counts to modulate brightness
normalized_iters = np.clip(iters_np / max_iterations, 0.0, 1.0)

# Hue (H): mapped according to the basin of attraction (root ID)
H = np.zeros_like(root_idx_np, dtype=np.float32)
H[root_idx_np == 1] = 0.05  # Root 1: Warm Orange/Red
H[root_idx_np == 2] = 0.35  # Root 2: Emerald Green
H[root_idx_np == 3] = 0.65  # Root 3: Deep Sky Blue

# Saturation (S): high saturation for vivid boundaries
S = np.ones_like(H) * 0.85

# Value/Brightness (V): faster convergence appears brighter
V = 1.0 - 0.85 * (normalized_iters**0.5)
V[root_idx_np == 0] = 0.0  # Unconverged points rendered as black

# Convert HSV representation to RGB image
rgb_image = hsv_to_rgb(np.stack([H, S, V], axis=-1))

# -------------------------------------------------------------------------
# Step 7: Plot and Display
# -------------------------------------------------------------------------
plt.figure(figsize=(10, 10))
plt.imshow(rgb_image)
plt.title(
    r"Newton Fractal for $f(z) = z^3 - 1$ (PyTorch Accelerated)", fontsize=14
)
plt.axis("off")
plt.tight_layout()
plt.show()
```
